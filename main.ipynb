```python
from huggingface_hub import hf_hub_download
```

```python
weights_path = hf_hub_download(
    repo_id="Samdutse/pothole-yolov8",
    filename="best.pt"
)

print(weights_path)
```

```python
from ultralytics import YOLO
```

```python
model = YOLO(weights_path)

results = model.predict(
    source="road.jpg",
    conf=0.25,
    save=True
)

results[0].show()
```
